### Problem 1: Employee Performance Bonus Eligibility
**Objective:**  
Identify employees with the highest performance score and handle ties.


```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Find the highest performance score
highest_score = max(employees.values())

# Find all employees with the highest score
top_performers = [name for name, score in employees.items() if score == highest_score]

print(f"Top Performers Eligible for Bonus: {', '.join(top_performers)} (Score: {highest_score})")
```

### Problem 2: Search Query Keyword Analysis
**Objective:**  
Analyze keyword frequency and display keywords searched more than once.


```python
query = "Buy mobile phone buy phone online"

# Convert to lowercase and split into words
words = query.lower().split()

# Count frequency of each word
frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

# Display only keywords with frequency greater than 1
repeated_keywords = {word: count for word, count in frequency.items() if count > 1}

print(repeated_keywords)
```

### Problem 3: Sensor Data Validation
**Objective:**  
Identify valid (even) sensor readings and store them with hour index.


```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid_readings = []

for index, value in enumerate(sensor_readings):
    if value % 2 == 0:
        valid_readings.append((index, value))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

### Problem 4: Email Domain Usage Analysis
**Objective:**  
Analyze email domain usage and calculate percentage distribution.


```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_count = {}

# Count domains
for email in emails:
    domain = email.split("@")[1]
    domain_count[domain] = domain_count.get(domain, 0) + 1

total_emails = len(emails)

# Calculate and display percentage
for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {int(percentage)}%")
```

### Problem 5: Sales Spike Detection
**Objective:**  
Detect days where sales exceed 30% above average.


```python
sales = [1200, 1500, 900, 2200, 1400, 3000]

average_sales = sum(sales) / len(sales)
threshold = average_sales * 1.3

for index, value in enumerate(sales):
    if value > threshold:
        print(f"Day {index + 1}: {value}")
```

### Problem 6: Duplicate User ID Detection
**Objective:**  
Identify duplicate user IDs and count their occurrences.


```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

id_count = {}

for user_id in user_ids:
    id_count[user_id] = id_count.get(user_id, 0) + 1

for user_id, count in id_count.items():
    if count > 1:
        print(f"{user_id} → {count} times")
```
