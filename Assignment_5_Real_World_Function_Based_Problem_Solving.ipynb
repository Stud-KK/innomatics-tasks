Problem 1: Social Media – Post Engagement Analyzer

Objective:
Analyze engagement on a social media post based on total likes.

```python
def post_eng(likes):
    tot = 0
    for l in likes:
        tot += l

    if tot >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:", tot)
    print("Post Status:", status)


post_eng([200, 350, 500])
```

Problem 3: Agriculture – Rainfall Adequacy Checker

```python
def rain_chk(data, req):
    total = 0
    for r in data:
        total += r

    avg = total // len(data)

    if avg >= req:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:", avg)
    print("Rainfall Status:", status)


rain_chk([70, 75, 72], 70)
```

Problem 4: Social Media – Duplicate Account Detection

```python
def dup_user(users):
    if len(users) != len(set(users)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


dup_user(["komal", "riya", "komal"])
```

Problem 6: Agriculture – Premium Crop Price Filter

```python
def prem_crop(prices):
    prem = []
    for p in prices:
        if p > 2000:
            prem.append(p)

    print("Premium Crops:", prem)


prem_crop([1500, 2500, 1800, 3200])
```

Problem 7: System Monitoring – Application Health Checker

```python
def app_health(err):
    if err == 0:
        status = "Healthy"
    elif err <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", err)
    print("System Status:", status)


app_health(7)
```

Problem 9: E-Learning – Student Attendance Eligibility System

```python
def att_chk(att):
    total = 0
    for a in att:
        total += a

    perc = (total / len(att))

    if perc >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print("Attendance Percentage:", perc)
    print("Exam Eligibility:", status)


att_chk([80, 75, 85])
```

Problem 10: Smart Electricity Bill Analyzer

```python
def ele_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7

    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    print("Total Bill:", bill)
    print("Usage Status:", usage)


ele_bill(250)
```
