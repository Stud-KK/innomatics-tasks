Problem 1: Smart Parking Lot Management System

```python
  def park_stat(cap, logs):
    cnt = 0

    for l in logs:
        if l == "IN":
            cnt += 1
        elif l == "OUT" and cnt > 0:
            cnt -= 1

    stat = "Full" if cnt >= cap else "Available"
    return cnt, stat


cap = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

p, s = park_stat(cap, logs)
print("Currently Parked Vehicles:", p)
print("Parking Status:", s)
```

Problem 2: Online Food Delivery Time Estimator

```python
def del_time(dist, traf, wea):
    t = dist * 5

    if traf == "High":
        t += 15
    elif traf == "Medium":
        t += 8

    if wea == "Rainy":
        t += 10
    elif wea == "Stormy":
        t += 20

    return t


d = 8
traf = "High"
wea = "Rainy"

print("Estimated Delivery Time:", del_time(d, traf, wea), "minutes")
```

Problem 3: Movie Theatre Seat Occupancy Analyzer

```python
def seat_stat(tot, book):
    b = len(book)
    per = (b / tot) * 100

    if per == 100:
        st = "Housefull"
    elif per >= 70:
        st = "Almost Full"
    else:
        st = "Available"

    return per, st


tot = 200
book = [1] * 150

o, s = seat_stat(tot, book)
print("Occupancy:", int(o), "%")
print("Show Status:", s)
```

Problem 4: Cloud Server Load Classification System

```python
def cpu_stat(read):
    avg = sum(read) / len(read)

    if avg < 50:
        st = "Normal"
    elif avg <= 80:
        st = "Warning"
    else:
        st = "Critical"

    return avg, st


read = [45, 60, 70, 85, 90]

a, s = cpu_stat(read)
print("Average CPU Load:", int(a), "%")
print("Server Status:", s)
```

Problem 5: Smart Classroom Resource Usage Monitor

```python
def res_stat(use):
    ov = []

    for r, h in use.items():
        if h > 8:
            ov.append(r)

    al = "Yes" if ov else "No"
    return ov, al


use = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

o, a = res_stat(use)
print("Overused Resources:", ", ".join(o))
print("Energy Alert:", a)
```

Problem 6: Online Event Registration Capacity Controller

```python
def reg_stat(cap, reg):
    con = cap if reg >= cap else reg
    wait = reg - cap if reg > cap else 0
    st = "Closed" if reg >= cap else "Open"

    return con, wait, st


cap = 100
reg = 105

c, w, s = reg_stat(cap, reg)
print("Confirmed Registrations:", c)
print("Waitlisted Users:", w)
print("Registration Status:", s)
```
